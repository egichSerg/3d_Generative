```python
import torch
import torch.nn as nn
```

```python
class Model(nn.Module):
    def __init__(self):
        super().__init__()

        self.fc = nn.Sequential(
            nn.Linear(in_features = 2, out_features = 5),
            nn.ReLU(),
            nn.Linear(in_features = 5, out_features = 10)
        )

    def forward(self, x):
        return self.fc(x)

model = Model()
```

```python
test_tensor = torch.randn(10, 5, 4, 6, 7, 8, 2)
test_tensor = torch.randn(6, 7, 8, 2)

output = model(test_tensor)
output.shape
```
