```python
from os import listdir
from os.path import isfile, isdir, join
from pathlib import Path

import numpy as np

mypath = Path('/home/student/work/3d_Generative/model/dataset/dataset')

classes = [d for d in listdir(mypath) if isdir(mypath / d) and d[0] != '.']
splits = {class_: split for class_, split in  zip(classes, [listdir(mypath / d) for d in classes])}
objects = {class_: {split_: objects for split_, objects in zip(split, [[val for val in listdir(mypath / class_ / split_name) if val[0] != '.' ] for split_name in split])} for class_, split in zip(splits.keys(), splits.values())}
objects
```

```python
import pandas as pd

def get_colors(path_to_filedir):
    return [i for i in listdir(path_to_filedir) if i.split('_')[-1] == 'rgb.pt'][0]

def get_densities(path_to_filedir):
    return [i for i in listdir(path_to_filedir) if i.split('_')[-1] == 'd.pt'][0]

dataframe_dict = {'object_id': [], 'class': [], 'split': [], 'color_path': [], 'density_path': []}
for class_ in objects.keys():
    for split in objects[class_].keys():
        for object_ in objects[class_][split]:
            dataframe_dict['object_id'].append(object_)
            dataframe_dict['class'].append(class_)
            dataframe_dict['split'].append(split)
            dataframe_dict['color_path'].append(mypath / class_ / split / object_ / get_colors(mypath / class_ / split / object_))
            dataframe_dict['density_path'].append(mypath / class_ / split / object_ / get_densities(mypath / class_ / split / object_))

data_real = pd.DataFrame(dataframe_dict)
data_real
```
